## 4. Factual Performance Analysis & Trade-Offs

### Factual Metric Performance:
1. **Random Forest & XGBoost**: Achieved the top factual performance across Accuracy, F1-Score, and ROC-AUC on both Binary and Multiclass tasks due to ensemble tree boosting handling non-linear interactions.
2. **Logistic Regression**: Achieved fast execution times but demonstrated lower Recall on complex attack categories due to linear decision boundaries.
3. **MLP Neural Network**: Demonstrated strong non-linear feature representation but required `max_iter=500` to guarantee complete loss convergence without early stopping warnings.

### Handling Training Issues & Class Imbalance:
- **MLP Convergence**: Increased `max_iter` to 500 and scaled input features with `StandardScaler` to ensure stable backpropagation gradients.
- **Multiclass Imbalance**: Weighted averaging (`average='weighted'`) and One-vs-Rest ROC-AUC were used so that minority attack categories (e.g. `Analysis`, `Backdoor`) contributed proportionally to overall system benchmarks.
