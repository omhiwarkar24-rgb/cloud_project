## 1. End-to-End Pipeline Execution Across 6 Operational Scenarios

We evaluate 6 comprehensive industrial scenarios covering the full spectrum of security and maintenance risk:

```python
scenarios = [
    {"case": "Scenario 1: Benign Normal Traffic + Healthy Engine", "attack": "Normal", "prob": 0.10},
    {"case": "Scenario 2: Reconnaissance Traffic + Moderate Wear", "attack": "Reconnaissance", "prob": 0.45},
    {"case": "Scenario 3: DoS Cyber Surge + Healthy Engine", "attack": "DoS", "prob": 0.15},
    {"case": "Scenario 4: Benign Traffic + High Failure Probability", "attack": "Normal", "prob": 0.85},
    {"case": "Scenario 5: Exploits Attack + High Failure Probability", "attack": "Exploits", "prob": 0.90},
    {"case": "Scenario 6: Worms Infection + Critical Hardware Failure", "attack": "Worms", "prob": 0.95}
]

pipeline_results = []
for s in scenarios:
    # Step 1: Risk Assessment
    risk_res = evaluate_integrated_risk(s["attack"], s["prob"])
    
    # Step 2: Prevention Recommendation
    prev_res = generate_recommendation(risk_res['security_risk_level'], risk_res['failure_risk_level'], s["prob"])
    
    pipeline_results.append({
        "Scenario": s["case"],
        "Attack_Category": s["attack"],
        "Security_Risk": risk_res['security_risk_level'],
        "Failure_Prob": f"{s['prob']:.0%}",
        "Failure_Risk": risk_res['failure_risk_level'],
        "Overall_System_Risk": f"{risk_res['overall_risk_level']} ({risk_res['overall_risk_score']})",
        "Action_Priority": prev_res['action_priority'],
        "Actionable_Recommendation": prev_res['recommendation']
    })

df_pipeline = pd.DataFrame(pipeline_results)
pd.set_option('display.max_colwidth', None)
display(df_pipeline)
```
