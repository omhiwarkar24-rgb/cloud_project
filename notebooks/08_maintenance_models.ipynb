## 3. Class Imbalance Handling Rationale & Performance Analysis

### Handling Class Imbalance:
- Machine failure represents only **7.70%** of samples (11.99:1 imbalance ratio). Standard unweighted classifiers would achieve ~92% accuracy by simply predicting all instances as Healthy (0), but would produce a **0% Recall** on actual machine breakdowns.
- To prevent this, we configured `class_weight='balanced'` for Logistic Regression, Decision Tree, and Random Forest, and `scale_pos_weight=12.0` for XGBoost. This heavily penalizes false negatives, forcing the models to prioritize identifying early failure signals.

### Factual Performance Trade-Offs:
1. **Random Forest & XGBoost**: Achieved top Recall and F1-Score on machine failure prediction by leveraging ensemble decision trees.
2. **Logistic Regression**: Achieved high Recall when weighted, but lower overall Precision due to linear decision boundaries.
3. **MLP Neural Network**: Demonstrated competitive ROC-AUC but required feature standardization to converge safely.
