## 1. Validation & Loading: UNSW-NB15 Cloud Security Dataset

We load `UNSW_NB15_training-set.csv` and `UNSW_NB15_testing-set.csv` from `data/security/`.

```python
security_dir = os.path.join(PROJECT_ROOT, 'data', 'security')
unsw_train_file = os.path.join(security_dir, 'UNSW_NB15_training-set.csv')
unsw_test_file = os.path.join(security_dir, 'UNSW_NB15_testing-set.csv')

# Validation Check
missing_sec_files = []
if not os.path.exists(unsw_train_file):
    missing_sec_files.append('UNSW_NB15_training-set.csv')
if not os.path.exists(unsw_test_file):
    missing_sec_files.append('UNSW_NB15_testing-set.csv')

if missing_sec_files:
    print(f"[DATASET VALIDATION ERROR]: Missing raw security dataset files: {missing_sec_files}")
    print(f"Please place the missing file(s) inside target folder: '{security_dir}'")
    print("Download link: https://research.unsw.edu.au/projects/unsw-nb15-dataset")
    sys.exit(1)
else:
    df_unsw_train = pd.read_csv(unsw_train_file)
    df_unsw_test = pd.read_csv(unsw_test_file)
    print("✅ UNSW-NB15 Training and Testing datasets successfully loaded!")
```

### UNSW-NB15 Training Set Inspection

```python
print(f"=== UNSW-NB15 Training Set Shape: {df_unsw_train.shape} ===")
print(f"Duplicate Rows: {df_unsw_train.duplicated().sum()}")
print("\n--- First 10 Rows ---")
display(df_unsw_train.head(10))

print("\n--- Column Names & Data Types ---")
print(df_unsw_train.dtypes)

print("\n--- Missing Values per Column ---")
missing_val_sec = df_unsw_train.isna().sum()
print(missing_val_sec[missing_val_sec > 0] if (missing_val_sec > 0).any() else "0 Missing Values Detected.")

print("\n--- Binary Target Distribution ('label') ---")
print(df_unsw_train['label'].value_counts(dropna=False))
print(df_unsw_train['label'].value_counts(normalize=True).map('{:.2%}'.format))

if 'attack_cat' in df_unsw_train.columns:
    print("\n--- Multi-class Attack Category Distribution ('attack_cat') ---")
    print(df_unsw_train['attack_cat'].value_counts(dropna=False))

print("\n--- Statistical Summary (describe) ---")
display(df_unsw_train.describe())
```

### UNSW-NB15 Testing Set Inspection

```python
print(f"=== UNSW-NB15 Testing Set Shape: {df_unsw_test.shape} ===")
print(f"Duplicate Rows: {df_unsw_test.duplicated().sum()}")
print("\n--- First 10 Rows ---")
display(df_unsw_test.head(10))

print("\n--- Target Distribution in Test Set ('label') ---")
print(df_unsw_test['label'].value_counts(dropna=False))
if 'attack_cat' in df_unsw_test.columns:
    print("\n--- Multi-class Attack Categories in Test Set ---")
    print(df_unsw_test['attack_cat'].value_counts(dropna=False))
```

## 2. Validation & Loading: AI4I 2020 Predictive Maintenance Dataset

We load `ai4i2020.csv` from `data/maintenance/`.

```python
maint_dir = os.path.join(PROJECT_ROOT, 'data', 'maintenance')
ai4i_file = os.path.join(maint_dir, 'ai4i2020.csv')

# Validation Check
if not os.path.exists(ai4i_file):
    print(f"[DATASET VALIDATION ERROR]: Missing raw predictive maintenance dataset file: 'ai4i2020.csv'")
    print(f"Please place the missing file inside target folder: '{maint_dir}'")
    print("Download link: https://archive.ics.uci.edu/dataset/601/ai4i")
    sys.exit(1)
else:
    df_ai4i = pd.read_csv(ai4i_file)
    print("✅ AI4I 2020 Predictive Maintenance dataset successfully loaded!")
```

### AI4I 2020 Dataset Inspection

```python
print(f"=== AI4I 2020 Dataset Shape: {df_ai4i.shape} ===")
print(f"Duplicate Rows: {df_ai4i.duplicated().sum()}")
print("\n--- First 10 Rows ---")
display(df_ai4i.head(10))

print("\n--- Column Names & Data Types ---")
print(df_ai4i.dtypes)

print("\n--- Missing Values per Column ---")
missing_val_maint = df_ai4i.isna().sum()
print(missing_val_maint[missing_val_maint > 0] if (missing_val_maint > 0).any() else "0 Missing Values Detected.")

print("\n--- Target Distribution ('Machine failure') ---")
if 'Machine failure' in df_ai4i.columns:
    print(df_ai4i['Machine failure'].value_counts(dropna=False))
    print(df_ai4i['Machine failure'].value_counts(normalize=True).map('{:.2%}'.format))

failure_modes = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']
present_modes = [col for col in failure_modes if col in df_ai4i.columns]
if present_modes:
    print("\n--- Breakdown of Specific Failure Modes ---")
    print(df_ai4i[present_modes].sum())

print("\n--- Statistical Summary (describe) ---")
display(df_ai4i.describe())
```
